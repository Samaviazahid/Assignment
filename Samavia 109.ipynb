```python
# Dictionary with student names as keys and grades as values
grades = {
    'Noor': 85,
    'Eman': 90,
    'Faseeha': 78,
    'Hussain': 92
}

# Calculate the average grade
average_grade = sum(grades.values()) / len(grades)

# Print the average grade
print(f"Average grade: {average_grade:.2f}")
```

```python
# Inventory
inventory = {
    'Tablet': {'quantity': 10, 'price': 500},
    'Earpods': {'quantity': 3, 'price': 100},
    'Phones': {'quantity': 2, 'price': 300}
}

# Update quantity
inventory['Phones']['quantity'] = 12

# Calculate total inventory value
total_value = sum(item['quantity'] * item['price'] for item in inventory.values())

# Output
print("Inventory:", inventory)
print("Total inventory value:", total_value)
print("Low-stock items:")
for item in inventory:
    if inventory[item]['quantity'] < 5:
        print(item, inventory[item])
```

```python
# Student Gradebook
students = {
    'Samavia': {'math': 92, 'english': 88, 'science': 78},
    'Noor': {'math': 86, 'english': 80, 'science': 92},
    'Rubab': {'math': 98, 'english': 96, 'science': 99}
}

# Calculate average grades for each student
averages = {name: sum(grades.values()) / len(grades) for name, grades in students.items()}

# Get math grades
math_grades = {name: grades['math'] for name, grades in students.items()}

# Find highest and lowest in math
highest_math = max(math_grades, key=math_grades.get)
lowest_math = min(math_grades, key=math_grades.get)

# Sort students by average grades in descending order
sorted_students = sorted(averages.items(), key=lambda x: x[1], reverse=True)

# Output
print("Average grades:", averages)
print("Highest math grade:", highest_math)
print("Lowest math grade:", lowest_math)
print("Students sorted by average grades:", sorted_students)
```
